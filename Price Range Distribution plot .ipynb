```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Dataset.csv')
df
```

```python
price_range=df['Price range'].value_counts()
print(price_range)
```

```python
sns.histplot(data=price_range,)
font1={'family':'serif','color':'green','size':20}
fontx={'family':'serif','color':'red','size':20}

fonty={'family':'serif','color':'blue','size':20}

plt.title('Distribution of Price Ranges', fontdict=font1)
plt.xlabel('Price Range', fontdict=fontx)
plt.ylabel('Number of Restaurants',fontdict=fonty)
plt.show()
```
